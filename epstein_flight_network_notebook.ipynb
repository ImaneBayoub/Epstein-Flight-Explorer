# Epstein Flight Network Visualization

Interactive visualization of flight records and co-travel networks.


## Overview

This project explores flight logs and passenger connections through an interactive D3.js visualization developed as part of a Master's degree (M2) in Data Science.

The visualization combines:
- a world map of flight routes,
- passenger co-travel relationships,
- interactive exploration of the data.
